```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.preprocessing import LabelEncoder

# 1. Create the dataset (Corrected to exactly 10 values each)
data = {
    'Distance': ['Short', 'Short', 'Short', 'Short', 'Long', 'Long', 'Long', 'Long', 'Short', 'Long'],
    'Weather':  ['Clear', 'Clear', 'Stormy', 'Stormy', 'Clear', 'Clear', 'Stormy', 'Stormy', 'Clear', 'Clear'], 
    'On_Time':  ['Yes', 'Yes', 'No', 'Yes', 'Yes', 'No', 'No', 'No', 'Yes', 'Yes']
}

df = pd.DataFrame(data)

# 2. Encode categorical data
le = LabelEncoder()
df_encoded = df.apply(le.fit_transform)

X = df_encoded[['Distance', 'Weather']]
y = df_encoded['On_Time']

# 3. Train Decision Tree
clf = DecisionTreeClassifier(criterion='gini', random_state=42)
clf.fit(X, y)

# 4. Print rules
tree_rules = export_text(clf, feature_names=['Distance', 'Weather'])
print("Decision Tree Rules:\n", tree_rules)
```
